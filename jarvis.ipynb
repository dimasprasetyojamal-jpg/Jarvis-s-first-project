```python
import datetime
import random
import os

# Memulihkan fungsi print bawaan jika sebelumnya tidak sengaja tertimpa
try:
    if isinstance(print, str):
        del print
except NameError:
    pass

# ==========================================
# 1. INISIALISASI DATA & MODUL
# ==========================================

# Penentuan Waktu Sapaan Otomatis
waktu_sekarang = datetime.datetime.now()
jam = waktu_sekarang.hour

if 4 <= jam < 11:
    sapaan_waktu = "Selamat pagi"
elif 11 <= jam < 15:
    sapaan_waktu = "Selamat siang"
elif 15 <= jam < 18:
    sapaan_waktu = "Selamat sore"
else:
    sapaan_waktu = "Selamat malam"

# Data Penyimpanan (Database Mini Jarvis)
catatan_tugas = []

kamus_cyber = {
    "firewall": "Sistem keamanan jaringan yang memantau dan mengontrol lalu lintas data berdasarkan aturan keamanan.",
    "phishing": "Teknik penipuan online untuk mengelabui pengguna agar memberikan data sensitif seperti kata sandi.",
    "malware": "Perangkat lunak berbahaya (seperti virus atau trojan) yang dirancang untuk merusak atau menyusup ke sistem.",
    "encryption": "Proses mengubah data terbaca menjadi sandi acak agar aman dan tidak bisa dibaca orang lain."
}

kumpulan_pesan = [
    "Motivasi: Konsistensi adalah kunci utama menguasai koding. Sedikit demi sedikit, lama-lama menjadi bukit!",
    "Fun Fact: Bahasa pemrograman Python dinamai bukan karena ular piton, tetapi dari acara komedi Inggris 'Monty Python's Flying Circus'!",
    "Motivasi: Jangan takut dengan error! Error adalah cara komputer memberi tahu kita bagian mana yang perlu diperbaiki.",
    "Fun Fact: Komputer pertama di dunia (ENIAC) berukuran sebesar ruangan besar dan memiliki berat hampir 30 ton.",
    "Motivasi: Menjadi programmer handal bukan tentang tidak pernah salah, tetapi tentang tidak pernah menyerah saat mencari solusinya."
]

contoh_soal = (
    "1. Berapa jumlah tangan manusia?",
    "2. Apa organ manusia terbesar?",
    "3. Apa perbedaan sel manusia dan hewan?"
)

daftar_menu = [
    "1. Cek Status Sistem",
    "2. Jalankan Pemindaian",
    "3. Lihat Jadwal Hari Ini",
    "4. Keluar",
    "5. Tanya Jawab Soal",
    "6. Kalkulator Pintar",
    "7. Generator Motivasi & Fun Fact",
    "8. Catatan Tugas Harian",
    "9. Kamus Istilah Cybersecurity",
    "10. Cari File di Direktori"  # <--- Fitur baru penelusur file!
]

# ==========================================
# 2. INTERAKSI AWAL / DIALOG
# ==========================================

print("Halo, biarkan saya memperkenalkan diriku. Nama saya Jarvis, saya adalah asisten pribadi Anda.")
nama = input("Sebelum itu, silakan masukkan nama Anda: ")

if nama.lower() == "tyo":
    print(f"{sapaan_waktu}, Tyo. Senang bisa terhubung kembali.")
    kabar = input("Bagaimana kabar hari ini, apakah baik atau buruk? ")

    if kabar.lower() == "buruk":
        print("Wah, kenapa bisa buruk? Mau kamu ceritakan atau tidak, kita akan bahas sama-sama.")
    else:
        print("Iya, saya senang mendengar kabar Anda baik-baik saja hari ini.")

    # ==========================================
    # 3. TAMPILAN MENU UTAMA
    # ==========================================
    print("\n--- MENU UTAMA JARVIS ---")
    for menu in daftar_menu:
        print(menu)

    pilihan = input("\nSilakan pilih nomor menu yang diinginkan: ")

    # ==========================================
    # 4. EKSEKUSI FITUR SESUAI PILIHAN
    # ==========================================

    if pilihan == "1":
        print("Status: Semua modul utama berjalan normal dan stabil.")
    elif pilihan == "2":
        print("Memindai direktori sistem... Tidak ada ancaman ditemukan.")
    elif pilihan == "3":
        print("Hari ini jadwal Anda fokus belajar pemrograman Python.")
    elif pilihan == "4":
        print("Terima kasih telah menggunakan Jarvis. Sampai jumpa!")

    elif pilihan == "5":
        print("Anda telah memilih fitur pemindaian jawaban. Mari kita bahas soal berikut:")
        for soal in contoh_soal:
            print(soal)

        pilih_soal = input("\nSilakan ketik nomor soal yang ingin dilihat jawabannya (1/2/3): ")

        if pilih_soal == "1":
            print("Jawaban: 2")
        elif pilih_soal == "2":
            print("Jawaban: Kulit")
        elif pilih_soal == "3":
            print("Jawaban: Sel manusia lebih lentur ketimbang sel tumbuhan karena sel manusia tidak memiliki dinding sel.")
        else:
            print("Nomor soal tidak valid.")

        tanya_lagi = input("\nApakah ada yang ingin ditanyakan? (iya/tidak): ")
        if tanya_lagi.lower() == "iya":
            respon_pengguna = input("Silakan, saya mendengarkan... ")
            if "ndk jadi" in respon_pengguna.lower() or "tidak jadi" in respon_pengguna.lower():
                print("eh ndk jadi deh hehe\nOk, selamat beristirahat!")
            else:
                print("Baik, saya mencatat pertanyaan Anda. Kita bahas setelah ini!")
        else:
            print("eh ndk jadi deh hehe\nOk, selamat beristirahat!")

    elif pilihan == "6":
        print("\n--- KALKULATOR PINTAR JARVIS ---")
        print("1. Penjumlahan (+)\n2. Pengurangan (-)\n3. Perkalian (*)\n4. Pembagian (/)")

        pilih_operasi = input("Masukkan pilihan operasi (1/2/3/4): ")
        angka1 = float(input("Masukkan angka pertama: "))
        angka2 = float(input("Masukkan angka kedua: "))

        if pilih_operasi == "1":
            print(f"Hasil: {angka1} + {angka2} = {angka1 + angka2}")
        elif pilih_operasi == "2":
            print(f"Hasil: {angka1} - {angka2} = {angka1 - angka2}")
        elif pilih_operasi == "3":
            print(f"Hasil: {angka1} * {angka2} = {angka1 * angka2}")
        elif pilih_operasi == "4":
            if angka2 != 0:
                print(f"Hasil: {angka1} / {angka2} = {angka1 / angka2}")
            else:
                print("Error: Tidak bisa membagi angka dengan nol (0)!")
        else:
            print("Pilihan operasi tidak valid.")

    elif pilihan == "7":
        print("\n--- GENERATOR MOTIVASI & FUN FACT JARVIS ---")
        print(random.choice(kumpulan_pesan))

    elif pilihan == "8":
        print("\n--- CATATAN Tugas HARIAN JARVIS ---")
        tugas_baru = input("Masukkan tugas baru yang ingin dicatat: ")
        catatan_tugas.append(tugas_baru)
        print(f"Berhasil! Tugas '{tugas_baru}' telah ditambahkan.")
        print("\nDaftar catatan tugasmu saat ini:")
        for index, tugas in enumerate(catatan_tugas, start=1):
            print(f"{index}. {tugas}")

    elif pilihan == "9":
        print("\n--- KAMUS ISTILAH CYBERSECURITY JARVIS ---")
        print("Daftar istilah yang tersedia: firewall, phishing, malware, encryption")
        kata_kunci = input("Masukkan istilah yang ingin dicari artinya: ").lower()

        if kata_kunci in kamus_cyber:
            print(f"\nDefinisi dari '{kata_kunci}':\n{kamus_cyber[kata_kunci]}")
        else:
            print(f"\nMaaf, istilah '{kata_kunci}' tidak ditemukan di dalam kamus Jarvis.")

    elif pilihan == "10":
        # --- Logika Fitur Pencari File ---
        print("\n--- PENCARI FILE JARVIS ---")
        kata_kunci_file = input("Masukkan nama file yang ingin dicari: ").lower()

        # Menggunakan direktori tempat script berjalan sebagai titik awal pencarian
        direktori_target = "."
        print(f"Memindai direktori untuk mencari file yang mengandung kata '{kata_kunci_file}'...")

        file_ditemukan = False
        # os.walk menelusuri folder dan sub-folder secara otomatis
        for root, dirs, files in os.walk(direktori_target):
            for file in files:
                if kata_kunci_file in file.lower():
                     jalur_lengkap = os.path.join(root, file)
                     print(f"[DITEMUKAN] -> {jalur_lengkap}")
                     file_ditemukan = True

        if not file_ditemukan:
            print("Pencarian selesai. Tidak ada file yang cocok dengan kata kunci tersebut.")

    else:
        print("Pilihan tidak dikenali, tetapi sistem tetap siaga.")

else:
    print("Akses ditolak. Identitas pengguna tidak dikenali.")
```
